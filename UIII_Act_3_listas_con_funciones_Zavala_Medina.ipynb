```python
db_streamers = [] # Lista global para almacenar los objetos Streamer [11]

def buscar_streamer(nickname):
    # Recorre la lista buscando una coincidencia de nombre [12, 13]
    for s in db_streamers:
        if s.nickname.lower() == nickname.lower():
            return s
    return None

def agregar_streamer():
    # Captura datos del usuario para crear un nuevo objeto [1, 14]
    print("\n--- Agregar Nuevo Streamer ---")
    nick = input("Nickname: ")
    if buscar_streamer(nick):
        print("Error: Ese streamer ya existe.")
        return
    plat = input("Plataforma: ")
    seg = int(input("Cantidad de seguidores: "))
    nuevo = Streamer(nick, plat, seg)
    db_streamers.append(nuevo) # Añade el objeto a la colección [15]
    print(f"¡{nick} ha sido agregado!")

def listar_streamers():
    # Muestra todos los elementos de la lista usando un ciclo for [12, 16]
    if not db_streamers:
        print("\nLa lista está vacía.")
    else:
        print("\n--- Lista de Streamers Registrados ---")
        for s in db_streamers:
            print(s)

def leer_streamer():
    # Busca y muestra la información detallada de un solo streamer [17, 18]
    nick = input("\nNickname a buscar: ")
    resultado = buscar_streamer(nick)
    if resultado:
        print(f"Datos encontrados: {resultado}")
    else:
        print("Streamer no encontrado.")

def modificar_streamer():
    # Localiza el objeto y actualiza sus atributos [19, 20]
    nick = input("\nNickname del streamer a modificar: ")
    s = buscar_streamer(nick)
    if s:
        print(f"Modificando a {nick}...")
        s.plataforma = input(f"Nueva plataforma (actual {s.plataforma}): ")
        s.seguidores = int(input(f"Nuevos seguidores (actual {s.seguidores}): "))
        print("Datos actualizados correctamente.")
    else:
        print("No se encontró el registro.")

def borrar_streamer():
    # Elimina el objeto de la lista [21, 22]
    nick = input("\nNickname del streamer a eliminar: ")
    s = buscar_streamer(nick)
    if s:
        db_streamers.remove(s)
        print(f"El streamer {nick} ha sido eliminado.")
    else:
        print("No se pudo eliminar: no existe.")


def ejecutar_sistema():
    while True:
        # Presenta las opciones al usuario [24, 25]
        print("\n=== SISTEMA DE GESTIÓN DE STREAMERS ===")
        print("1. Agregar streamer")
        print("2. Listar todos")
        print("3. Buscar/Leer uno")
        print("4. Modificar datos")
        print("5. Borrar registro")
        print("6. Salir")

        opcion = input("Elija una opción (1-6): ")

        if opcion == '1':
            agregar_streamer()
        elif opcion == '2':
            listar_streamers()
        elif opcion == '3':
            leer_streamer()
        elif opcion == '4':
            modificar_streamer()
        elif opcion == '5':
            borrar_streamer()
        elif opcion == '6':
            print("Saliendo del sistema...")
            print("Programa Realizado por Zavala Medina")
            break # Termina el ciclo y el programa [2, 26]
        else:
            print("Opción no válida, intente de nuevo.")

# Punto de entrada del programa [27]
if __name__ == "__main__":
    ejecutar_sistema()
```
